```python
from sklearn.datasets import fetch_california_housing
from sklearn.metrics import root_mean_squared_error
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPRegressor
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

housing = fetch_california_housing()
X_train, X_test, y_train, y_test = train_test_split(housing.data, housing.target, random_state=0)

regMLP = MLPRegressor([50,50,50],early_stopping = True,verbose = True,random_state = 0)
pipeline = make_pipeline(StandardScaler(),regMLP)
pipeline.fit(X_train,y_train)
```

```python
from sklearn.datasets import fetch_openml

fashion_mnist = fetch_openml(name = "Fashion-MNIST",as_frame = False)
target = fashion_mnist.target.astype(int)

X_train,y_train = fashion_mnist.data[:60000,target[:60000]]
X_test,y_test = fashion_mnist.data[60000:],target[:60000]
```

```python
from sklearn.neural_network import MLPClassifier
from sklearn.preprocessing import MinMaxScaler
from sklearn.pipeline import make_pipeline

mlp_clf = MLPClassifier(hidden_layer_sizes=[300,100],verbose = True,early_stopping=True,random_state=0)
pipeline = make_pipeline(MinMaxScaler(),mlp_clf)
pipeline.fit(X_train,y_train)
```
